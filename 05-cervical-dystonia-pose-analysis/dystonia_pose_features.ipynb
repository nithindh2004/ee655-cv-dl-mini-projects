```python
import os, pandas as pd
import dystonia_pose_features as dpf

dpf.ensure_model()
os.makedirs(dpf.OUT_DIR, exist_ok=True)

IMAGES = {  # path: true label
    "man1.jpg": "Normal",
    "man2.jpg": "Normal",
    "woman_rot.jpg": "Rot. Torticollis",
    "woman_lat.jpg": "Laterocollis",
}
```

```python
rows = []
for path, label in IMAGES.items():
    feats, ann = dpf.process_image(path)
    if feats is None:
        continue
    pred = dpf.classify_pose(feats)
    name = os.path.splitext(os.path.basename(path))[0]
    dpf.show_result(ann, feats, f"{label} | Predicted: {pred}", os.path.join(dpf.OUT_DIR, f"result_{name}.png"))
    rows.append({"Image": name, "True": label, "Predicted": pred, **feats})
df = pd.DataFrame(rows)
dpf.comparison_chart(df, os.path.join(dpf.OUT_DIR, "comparison_chart.png"))
df
```

**Reported:** F2 (nose offset) flags rotation (−0.359 vs ≤ 0.051 for normals); F3 + F4 flag laterocollis (0.279 and −17.7° vs ≤ 0.010 and ≤ 1.9°). The rule table classifies all four test images correctly.
