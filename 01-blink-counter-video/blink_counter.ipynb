# Per-eye blink counting from a patient video
Thin driver around `blink_counter.py` (kept as the single source of truth). Needs the Blink Detection Dataset and `patient.mp4`.

```python
# !pip install torch torchvision opencv-python mediapipe==0.10.14 pillow matplotlib
```

```python
import os
from blink_counter import (train_classifier, process_video, make_plots, print_report,
                           EyeClassifier, DEVICE)

DATA_DIR = "Blink Detection Dataset"
VIDEO = "patient.mp4"
OUT = "blink_output"
DETECTOR = "haar"   # or "mediapipe"
os.makedirs(OUT, exist_ok=True)
```

### Phase 1 — train the open/closed classifier

```python
model, val_acc = train_classifier(DATA_DIR, OUT)
```

### Phase 2 — crop eyes, classify every frame, count blinks

```python
left, right, fps, n_frames, face_frames, samples = process_video(model, VIDEO, OUT, DETECTOR)
make_plots(left, right, fps, OUT, samples)
print_report(left, right, fps, n_frames, face_frames, val_acc)
```
